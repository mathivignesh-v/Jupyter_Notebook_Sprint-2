## T-Test

### Step 1 – Understand the Concept
Compares means using a t distribution; Welch's version is useful when equal variances cannot be assumed.

### Step 2 – Demonstrate the Concept
Compare average customer spending in two independent groups.

**Assumptions:** Independent observations; suitable normality/large-sample conditions; equal variance only for pooled t-test.

**When to use:** Use for suitable comparisons of two means.

**When not to use:** Do not use a standard independent t-test for paired/dependent data.

**Business Example:** Compare average customer spending between two independent segments.

```python
from scipy.stats import ttest_ind
a=[120,130,125,140,135,128]; b=[150,155,148,160,152,158]
print(ttest_ind(a,b,equal_var=False))
```

### Interpretation
A small p-value provides evidence that the group means differ under the assumptions.

### Key Takeaway
Use the t-test for suitable two-group mean comparisons.

## Z-Test

### Step 1 – Understand the Concept
Uses a standard normal reference distribution, commonly when population SD is known or a justified large-sample z approximation applies.

### Step 2 – Demonstrate the Concept
Test a process mean against a target when population SD is known.

**Assumptions:** Known population SD or justified large-sample approximation and appropriate independence/design.

**When to use:** Use when its known-SD/large-sample framework is justified.

**When not to use:** Do not assume z-testing is valid merely because n is large.

**Business Example:** Check whether product weight differs from a regulated target with known process SD.

```python
from scipy.stats import norm
x=np.array([101,99,100,102,98,101,100,99,102,100])
mu0=100; sigma=2
z=(x.mean()-mu0)/(sigma/np.sqrt(len(x)))
p=2*(1-norm.cdf(abs(z)))
print(z,p)
```

### Interpretation
The p-value assesses evidence against the target mean under the stated model.

### Key Takeaway
Use a z framework only when its assumptions/design are justified.

## Chi-Square Test

### Step 1 – Understand the Concept
Tests categorical count relationships, including independence and goodness-of-fit.

### Step 2 – Demonstrate the Concept
Test whether membership tier is associated with payment method.

**Assumptions:** Independent categorical observations and adequate expected cell counts.

**When to use:** Use for categorical association or goodness-of-fit.

**When not to use:** Do not ignore sparse expected counts or dependence.

**Business Example:** Test association between customer tier and payment method.

```python
from scipy.stats import chi2_contingency
table=np.array([[30,20,10],[25,25,15]])
print(chi2_contingency(table))
```

### Interpretation
A small p-value suggests an association between the categorical variables.

### Key Takeaway
Check expected counts and independence assumptions.

## ANOVA

### Step 1 – Understand the Concept
Tests whether three or more independent group means are all equal.

### Step 2 – Demonstrate the Concept
Compare average sales across three regions.

**Assumptions:** Independent observations, suitable residual normality, and homogeneity of variance for classical one-way ANOVA.

**When to use:** Use for comparing three or more independent group means.

**When not to use:** Do not use classical one-way ANOVA blindly for repeated measures or severe assumption violations.

**Business Example:** Compare average sales across three regions.

```python
from scipy.stats import f_oneway
north=[100,105,98,110,103]; south=[120,125,118,122,127]; west=[110,108,115,112,117]
print(f_oneway(north,south,west))
```

### Interpretation
A significant result means at least one mean differs; follow-up comparisons are needed.

### Key Takeaway
ANOVA is an omnibus test, not a complete pairwise explanation.

## Mann-Whitney U Test

### Step 1 – Understand the Concept
A rank-based nonparametric test for two independent groups.

### Step 2 – Demonstrate the Concept
Compare customer satisfaction ratings between two channels.

**Assumptions:** Independent observations and ordinal/continuous outcome; interpretation depends on distribution shapes.

**When to use:** Use for two independent groups when a rank-based method is appropriate.

**When not to use:** Do not describe it simply as a test of means.

**Business Example:** Compare satisfaction scores between two service channels.

```python
from scipy.stats import mannwhitneyu
a=[3,4,4,5,2,4]; b=[2,3,3,4,2,3]
print(mannwhitneyu(a,b,alternative="two-sided"))
```

### Interpretation
A small p-value indicates a difference in rank distributions under the test assumptions.

### Key Takeaway
Use it when a rank-based comparison is appropriate.

## Shapiro-Wilk Test

### Step 1 – Understand the Concept
Tests whether a sample is consistent with a normal distribution.

### Step 2 – Demonstrate the Concept
Check residuals as one part of a normality diagnostic.

**Assumptions:** Independent observations; continuous sample; interpretation depends on sample size.

**When to use:** Use as one normality diagnostic.

**When not to use:** Do not treat a non-significant result as proof of perfect normality.

**Business Example:** Check residual distribution before an analysis requiring approximate normality.

```python
from scipy.stats import shapiro
x=np.array([-1.1,-.4,.2,.5,.1,-.2,.7,-.1,.3,0])
print(shapiro(x))
```

### Interpretation
A small p-value provides evidence against normality; plots and domain knowledge should also be considered.

### Key Takeaway
Normality testing is a diagnostic, not a guarantee of model validity.

## Test Selection Guide

| Question | Typical Test |
|---|---|
| Two independent means | T-test / Welch T-test |
| Known SD / z framework | Z-test |
| Categorical association | Chi-square |
| Three or more independent means | ANOVA |
| Two independent rank-based groups | Mann-Whitney U |
| Normality diagnostic | Shapiro-Wilk |

Statistical tests should be selected from the research question, data type, study design, and assumptions.
