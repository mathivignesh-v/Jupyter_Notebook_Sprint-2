# Mean

## Step 1 – Understand the Concept

### Definition
The mean is the arithmetic average of all observations.

### Formula
**Mean = Sum of all observations / Number of observations**


## Step 2 – Demonstrate the Concept

### Numerical Example
10, 20, 30, 40, 50 → Mean = 150 / 5 = 30

### Business Example
The mean can summarize average monthly sales or average customer spending.

### AI/ML Example
The mean can summarize a numerical feature and can be used for mean-based missing-value imputation.

The examples show how the measure can represent or summarize real-world data.


## Step 3 – Implement the Concept

The following Python program demonstrates the concept with a small dataset.


```python
import numpy as np

data = np.array([10, 20, 30, 40, 50])
print("Mean:", np.mean(data))
```

## Interpretation

The mean is 30. It represents the arithmetic center of the data but can be affected by extreme values.

### Key Takeaway
The important part is not only calculating the value, but understanding what it represents and when it should be used.


# Median

## Step 1 – Understand the Concept

### Definition
The median is the middle value after arranging observations in order. With an even number of values, it is the average of the two middle values.

### Formula
**For odd n: middle value. For even n: average of the two middle values.**


## Step 2 – Demonstrate the Concept

### Numerical Example
10, 20, 30, 40, 50 → Median = 30. For 10, 20, 30, 40 → Median = (20+30)/2 = 25.

### Business Example
Median salary can represent a typical employee salary when a few very high salaries would distort the mean.

### AI/ML Example
The median is useful for understanding skewed features and for median-based missing-value imputation.

The examples show how the measure can represent or summarize real-world data.


## Step 3 – Implement the Concept

The following Python program demonstrates the concept with a small dataset.


```python
import numpy as np

data = np.array([10, 20, 30, 40, 50])
print("Median:", np.median(data))
```

## Interpretation

The median is 30. It is generally less affected by extreme values than the mean.

### Key Takeaway
The important part is not only calculating the value, but understanding what it represents and when it should be used.


# Mode

## Step 1 – Understand the Concept

### Definition
The mode is the value that occurs most frequently in a dataset.

### Formula
**Mode = Observation with the highest frequency.**


## Step 2 – Demonstrate the Concept

### Numerical Example
2, 3, 3, 4, 5, 3, 6 → Mode = 3 because 3 occurs most often.

### Business Example
A retailer can find the most commonly purchased product size or category.

### AI/ML Example
Mode is useful for categorical data and can be used to identify the most common category.

The examples show how the measure can represent or summarize real-world data.


## Step 3 – Implement the Concept

The following Python program demonstrates the concept with a small dataset.


```python
from collections import Counter

data = [2, 3, 3, 4, 5, 3, 6]
frequency = Counter(data)
mode = frequency.most_common(1)[0][0]

print("Frequency:", frequency)
print("Mode:", mode)
```

## Interpretation

The mode is 3. It is especially useful when averaging categories is not meaningful.

### Key Takeaway
The important part is not only calculating the value, but understanding what it represents and when it should be used.


# Weighted Mean

## Step 1 – Understand the Concept

### Definition
A weighted mean gives different observations different levels of importance.

### Formula
**Weighted Mean = Σ(value × weight) / Σ(weights)**


## Step 2 – Demonstrate the Concept

### Numerical Example
Marks 80, 70, 90 with weights 0.2, 0.3, 0.5 → Weighted Mean = 82.

### Business Example
A company can calculate an employee score where sales, quality, and attendance have different weights.

### AI/ML Example
Weighted averages can be useful in evaluation metrics, survey analysis, sampling, and other ML-related calculations.

The examples show how the measure can represent or summarize real-world data.


## Step 3 – Implement the Concept

The following Python program demonstrates the concept with a small dataset.


```python
import numpy as np

marks = np.array([80, 70, 90])
weights = np.array([0.2, 0.3, 0.5])

print("Weighted Mean:", np.average(marks, weights=weights))
```

## Interpretation

The result is 82. The value 90 has the greatest influence because it has the largest weight.

### Key Takeaway
The important part is not only calculating the value, but understanding what it represents and when it should be used.


# Geometric Mean

## Step 1 – Understand the Concept

### Definition
The geometric mean is an average based on multiplication. It is useful for positive values involving growth rates, ratios, or multiplicative changes.

### Formula
**Geometric Mean = (x₁ × x₂ × ... × xₙ)^(1/n)**


## Step 2 – Demonstrate the Concept

### Numerical Example
2, 8, 32 → Geometric Mean = (2×8×32)^(1/3) = 8.

### Business Example
Geometric mean can summarize average investment growth over several periods.

### AI/ML Example
It can be useful for multiplicative quantities, growth factors, and ratios.

The examples show how the measure can represent or summarize real-world data.


## Step 3 – Implement the Concept

The following Python program demonstrates the concept with a small dataset.


```python
import numpy as np

data = np.array([2, 8, 32], dtype=float)
geometric_mean = np.prod(data) ** (1 / len(data))

print("Geometric Mean:", geometric_mean)
```

## Interpretation

The geometric mean is 8. It represents a central multiplicative level and is normally used with positive values.

### Key Takeaway
The important part is not only calculating the value, but understanding what it represents and when it should be used.


# Harmonic Mean

## Step 1 – Understand the Concept

### Definition
The harmonic mean is calculated using reciprocals and is particularly useful for rates and ratios.

### Formula
**Harmonic Mean = n / Σ(1/xᵢ)**


## Step 2 – Demonstrate the Concept

### Numerical Example
30 and 60 → Harmonic Mean = 2 / (1/30 + 1/60) = 40.

### Business Example
For equal-distance travel at different speeds, harmonic mean can represent the average speed more appropriately.

### AI/ML Example
In ML, the harmonic mean is used in metrics such as the F1-score, which combines precision and recall.

The examples show how the measure can represent or summarize real-world data.


## Step 3 – Implement the Concept

The following Python program demonstrates the concept with a small dataset.


```python
import numpy as np

rates = np.array([30, 60], dtype=float)
harmonic_mean = len(rates) / np.sum(1 / rates)

print("Harmonic Mean:", harmonic_mean)
```

## Interpretation

The harmonic mean is 40. It gives more influence to smaller values than the arithmetic mean, making it useful for certain rate-based problems.

### Key Takeaway
The important part is not only calculating the value, but understanding what it represents and when it should be used.


# Comparison of Central Tendency Measures

## 1. Practical Problem – Comparing Mean, Median and Mode

A dataset contains daily transaction values:

`[20, 22, 21, 23, 24, 25, 22, 21, 23, 22, 150]`

The value 150 may represent an unusually large transaction. Calculate the mean, median, and mode and determine which measure best represents a typical transaction.

```python
import numpy as np
import matplotlib.pyplot as plt
from collections import Counter

transactions = np.array([20, 22, 21, 23, 24, 25, 22, 21, 23, 22, 150])

mean_value = transactions.mean()
median_value = np.median(transactions)
mode_value = Counter(transactions).most_common(1)[0][0]

print("Mean:", mean_value)
print("Median:", median_value)
print("Mode:", mode_value)

plt.figure(figsize=(8, 4))
plt.hist(transactions, bins=8, edgecolor="black")
plt.axvline(mean_value, linestyle="--", label=f"Mean = {mean_value:.2f}")
plt.axvline(median_value, linestyle="-", label=f"Median = {median_value:.2f}")
plt.axvline(mode_value, linestyle=":", label=f"Mode = {mode_value}")
plt.title("Transaction Values with Central Tendency Measures")
plt.xlabel("Transaction value")
plt.ylabel("Frequency")
plt.legend()
plt.show()
```

### Interpretation

The extreme value of 150 pulls the mean upward, while the median remains closer to the typical transaction value. The mode identifies the most frequently occurring value. For skewed data with outliers, the median can be a more representative measure of central tendency than the mean.

### AI/ML Relevance

Central tendency is useful during EDA and feature analysis. Engineers compare mean and median to identify skewness, understand typical values, and decide whether transformations or robust statistics are appropriate.
