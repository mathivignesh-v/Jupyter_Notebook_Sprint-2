# 1. Normal Distribution

## Step 1 – Understand the Concept

### Definition
A normal distribution is a continuous, bell-shaped distribution that is symmetric around its mean. Values near the mean are more common than values far from it.

### Formula / Key Idea
For a normal distribution, the probability density is determined by its mean μ and standard deviation σ.


## Step 2 – Demonstrate the Concept

### Numerical Example
A normal distribution with mean 50 and standard deviation 10 is centered at 50, with most values concentrated around the center.

### Business Example
Test scores or measurement errors may approximately follow a normal distribution in some situations.

### AI/ML Use Case
Normality is useful to understand feature distributions and underlies many statistical methods, although many ML algorithms do not require features to be normally distributed.

This example demonstrates how the concept helps us understand the spread or distribution of data.


## Step 3 – Implement the Concept

The following Python code demonstrates the concept.

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)
data = rng.normal(loc=50, scale=10, size=1000)
plt.hist(data, bins=30)
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.title('Normal Distribution')
plt.show()
```

## Interpretation

The histogram should show a roughly bell-shaped pattern centered near 50.

### Key Takeaway
The goal is to understand what the result means and when this concept is useful, not simply to calculate a value.


# 2. Uniform Distribution

## Step 1 – Understand the Concept

### Definition
A uniform distribution gives values in a specified interval approximately equal probability density.

### Formula / Key Idea
For a continuous uniform distribution on [a,b], density = 1/(b−a) between a and b.


## Step 2 – Demonstrate the Concept

### Numerical Example
For values uniformly distributed from 0 to 10, every interval of equal length within 0–10 has the same probability.

### Business Example
A simulation may use a uniform distribution when every value in a defined range is considered equally likely.

### AI/ML Use Case
Uniform random values are often used in simulation and initialization or sampling procedures.

This example demonstrates how the concept helps us understand the spread or distribution of data.


## Step 3 – Implement the Concept

The following Python code demonstrates the concept.

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)
data = rng.uniform(0, 10, 1000)
plt.hist(data, bins=20)
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.title('Uniform Distribution')
plt.show()
```

## Interpretation

The histogram should appear relatively flat across the interval 0 to 10.

### Key Takeaway
The goal is to understand what the result means and when this concept is useful, not simply to calculate a value.


# 3. Gaussian Distribution

## Step 1 – Understand the Concept

### Definition
Gaussian distribution is another name commonly used for the normal distribution. It is characterized by a mean and standard deviation and has a symmetric bell shape.

### Formula / Key Idea
Gaussian density uses mean μ and standard deviation σ to describe the center and spread.


## Step 2 – Demonstrate the Concept

### Numerical Example
For μ = 0 and σ = 1, the standard Gaussian distribution is centered at zero with standard deviation one.

### Business Example
Measurement errors can sometimes be modeled approximately using a Gaussian distribution.

### AI/ML Use Case
Gaussian assumptions appear in statistical modeling, Gaussian Naive Bayes, noise models, and many mathematical techniques.

This example demonstrates how the concept helps us understand the spread or distribution of data.


## Step 3 – Implement the Concept

The following Python code demonstrates the concept.

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-4, 4, 500)
mu = 0
sigma = 1
pdf = (1 / (sigma * np.sqrt(2 * np.pi))) * np.exp(-0.5 * ((x - mu) / sigma) ** 2)
plt.plot(x, pdf)
plt.xlabel('x')
plt.ylabel('Density')
plt.title('Gaussian Distribution')
plt.show()
```

## Interpretation

The curve is centered at zero and is symmetric, with density decreasing as values move away from the mean.

### Key Takeaway
The goal is to understand what the result means and when this concept is useful, not simply to calculate a value.


# 4. Left Skew

## Step 1 – Understand the Concept

### Definition
A left-skewed distribution has a longer tail toward smaller values. It is also called negatively skewed.

### Formula / Key Idea
Key idea: the distribution has a longer tail on the left side; typically mean < median in a strongly left-skewed distribution.


## Step 2 – Demonstrate the Concept

### Numerical Example
Exam scores where most students score high but a small number score very low can produce left skew.

### Business Example
If most customers spend large amounts but a few spend very little, spending data may show left skew.

### AI/ML Use Case
Recognizing skew can guide transformations and help an engineer choose suitable summaries and preprocessing methods.

This example demonstrates how the concept helps us understand the spread or distribution of data.


## Step 3 – Implement the Concept

The following Python code demonstrates the concept.

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)
data = 100 - rng.exponential(scale=10, size=1000)
data = data[data >= 0]
plt.hist(data, bins=30)
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.title('Left-Skewed Distribution')
plt.show()
```

## Interpretation

The longer tail extends toward smaller values. The mean can be pulled toward that tail.

### Key Takeaway
The goal is to understand what the result means and when this concept is useful, not simply to calculate a value.


# 5. Right Skew

## Step 1 – Understand the Concept

### Definition
A right-skewed distribution has a longer tail toward larger values. It is also called positively skewed.

### Formula / Key Idea
Key idea: the distribution has a longer tail on the right side; typically mean > median in a strongly right-skewed distribution.


## Step 2 – Demonstrate the Concept

### Numerical Example
Income data often has many moderate values and a small number of very high incomes, producing right skew.

### Business Example
Customer purchase amounts or income can be right-skewed because a small number of customers may spend much more than most customers.

### AI/ML Use Case
Right-skewed features may benefit from transformations such as logarithms when appropriate, depending on the model and business meaning.

This example demonstrates how the concept helps us understand the spread or distribution of data.


## Step 3 – Implement the Concept

The following Python code demonstrates the concept.

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)
data = rng.exponential(scale=20, size=1000)
plt.hist(data, bins=30)
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.title('Right-Skewed Distribution')
plt.show()
```

## Interpretation

The long right tail indicates a smaller number of unusually large observations.

### Key Takeaway

# 6. Symmetric Distribution

## Step 1 – Understand the Concept

### Definition
A symmetric distribution has similar shape on both sides of its center. The left and right sides mirror each other around the center.

### Formula / Key Idea
Key idea: the distribution has approximately equal shape on both sides of its central value.


## Step 2 – Demonstrate the Concept

### Numerical Example
A normal distribution centered at 50 is symmetric because values equally far above and below 50 have matching positions in the distribution.

### Business Example
A measurement process with errors that are equally likely to be above or below the target can show a symmetric distribution.

### AI/ML Use Case
Understanding symmetry helps with distribution analysis and deciding whether transformations or robust statistics may be useful.

This example demonstrates how the concept helps us understand the spread or distribution of data.


## Step 3 – Implement the Concept

The following Python code demonstrates the concept.

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-4, 4, 500)
pdf = np.exp(-0.5 * x**2) / np.sqrt(2 * np.pi)
plt.plot(x, pdf)
plt.xlabel('Value')
plt.ylabel('Density')
plt.title('Symmetric Distribution')
plt.show()
```

## Interpretation

The two sides have similar shapes around the center, so there is no long tail favoring one direction.

### Key Takeaway
The goal is to understand what the result means and when this concept is useful, not simply to calculate a value.


# 7. Kurtosis

## Step 1 – Understand the Concept

### Definition
Kurtosis describes aspects of the tail behavior of a distribution relative to a reference distribution. In practical data analysis, it helps identify whether extreme observations are more or less prominent than expected under a normal reference.

### Formula / Key Idea
Excess kurtosis is commonly reported as kurtosis − 3 for a normal reference, so a normal distribution has excess kurtosis near 0.


## Step 2 – Demonstrate the Concept

### Numerical Example
If a dataset has excess kurtosis near 0, its tail behavior is similar to the normal reference under that convention. Positive excess kurtosis indicates heavier tails, while negative indicates lighter tails.

### Business Example
Kurtosis can help identify whether business data contains unusually frequent extreme observations, such as very large transaction amounts.

### AI/ML Use Case
Tail behavior matters in anomaly detection, risk analysis, and deciding how to handle extreme observations.

This example demonstrates how the concept helps us understand the spread or distribution of data.


## Step 3 – Implement the Concept

The following Python code demonstrates the concept.

```python
import numpy as np
from scipy.stats import kurtosis

rng = np.random.default_rng(42)
data = rng.normal(size=1000)

print('Excess kurtosis:', kurtosis(data, fisher=True))
```

## Interpretation

A positive excess kurtosis suggests heavier tails and potentially more extreme observations than a normal reference; negative values suggest lighter tails.

### Key Takeaway
The goal is to understand what the result means and when this concept is useful, not simply to calculate a value.


# 8. Empirical Rule (68–95–99.7)

## Step 1 – Understand the Concept

### Definition
For an approximately normal distribution, the empirical rule states that about 68% of observations lie within one standard deviation of the mean, about 95% within two, and about 99.7% within three.

### Formula / Key Idea
Approximately 68%: μ ± 1σ; 95%: μ ± 2σ; 99.7%: μ ± 3σ.


## Step 2 – Demonstrate the Concept

### Numerical Example
If mean = 100 and standard deviation = 10, about 68% of observations are expected between 90 and 110, about 95% between 80 and 120, and about 99.7% between 70 and 130, assuming normality.

### Business Example
A company can use the rule to identify values that are unusually far from the average when a metric is approximately normal.

### AI/ML Use Case
The rule helps interpret standard deviations and identify potentially unusual observations in approximately normal data.

This example demonstrates how the concept helps us understand the spread or distribution of data.


## Step 3 – Implement the Concept

The following Python code demonstrates the concept.

```python
import numpy as np

mu = 100
sigma = 10

for k in [1, 2, 3]:
    lower = mu - k * sigma
    upper = mu + k * sigma
    print(f'{k} SD: {lower} to {upper}')
```

## Interpretation

The intervals become wider as the number of standard deviations increases. The rule is an approximation and should not be applied blindly to skewed or non-normal data.

### Key Takeaway
The goal is to understand what the result means and when this concept is useful, not simply to calculate a value.
