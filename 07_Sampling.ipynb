# 1. Sampling

## Step 1 – Understand the Concept

### Definition
Sampling is the process of selecting a subset of observations from a larger population.

Population = complete group of interest.
Sample = selected subset.

Sampling is useful when studying the entire population is expensive, slow, or impractical.

## Step 2 – Demonstrate the Concept

### Numerical Example
Population = 1,000 customers.
Sample = 100 selected customers.

### Business Example
A company may survey 500 customers instead of all customers.

### AI/ML Use Case
A representative sample can be used for exploratory analysis or preliminary model development before processing a very large dataset.

This example demonstrates how the concept can be understood and applied.

## Step 3 – Implement the Concept

The following Python code demonstrates the concept.

```python
import pandas as pd

population = pd.DataFrame({
    "customer_id": range(1, 101)
})

sample = population.sample(n=10, random_state=42)

print(sample)
```

## Interpretation

Ten observations are selected from a population of 100.

A good sampling method should preserve important characteristics of the population.

### Key Takeaway
The goal is to understand what the result means and when this concept is useful, not simply to calculate a value.

# 2. Random Sampling

## Step 1 – Understand the Concept

### Definition
Random sampling selects observations randomly so that population members have a known chance of being selected.

It reduces deliberate selection and convenience-based sampling.

## Step 2 – Demonstrate the Concept

### Numerical Example
From 1,000 customers, randomly select 100 customers.

### Business Example
Randomly select customers for a satisfaction survey.

### AI/ML Use Case
Random train/test splitting can be appropriate when observations are independent and similarly distributed.

This example demonstrates how the concept can be understood and applied.

## Step 3 – Implement the Concept

The following Python code demonstrates the concept.

```python
import numpy as np

rng = np.random.default_rng(42)

population = np.arange(1, 101)
sample = rng.choice(population, size=10, replace=False)

print("Random sample:", sample)
```

## Interpretation

The selected observations are chosen without manually deciding which customers should be included.

Random sampling can reduce selection bias, but rare subgroups may still be underrepresented by chance.

### Key Takeaway
The goal is to understand what the result means and when this concept is useful, not simply to calculate a value.

# 3. Stratified Sampling

## Step 1 – Understand the Concept

### Definition
Stratified sampling divides the population into meaningful subgroups called strata and samples from each subgroup.

The purpose is to ensure important groups are represented.

## Step 2 – Demonstrate the Concept

### Numerical Example
Suppose customers belong to Regions A, B, and C. A stratified sample takes observations from all three regions.

### Business Example
A company can sample customers separately from different age groups or regions.

### AI/ML Use Case
Stratified train/test splitting helps preserve class proportions, especially in imbalanced classification.

This example demonstrates how the concept can be understood and applied.

## Step 3 – Implement the Concept

The following Python code demonstrates the concept.

```python
import pandas as pd

data = pd.DataFrame({
    "customer_id": range(1, 13),
    "region": ["A"]*5 + ["B"]*4 + ["C"]*3
})

sample = data.groupby(
    "region", group_keys=False
).sample(n=1, random_state=42)

print(sample)
```

## Interpretation

Each region contributes to the final sample.

This is useful when subgroup representation matters.

### Key Takeaway
The goal is to understand what the result means and when this concept is useful, not simply to calculate a value.

# 4. Cluster Sampling

## Step 1 – Understand the Concept

### Definition
Cluster sampling divides the population into natural groups called clusters and selects one or more clusters for study.

Examples:
- Schools
- Stores
- Villages
- Branches
- Geographic regions

## Step 2 – Demonstrate the Concept

### Numerical Example
A company has 100 stores. Instead of sampling customers from every store, it selects 10 stores and studies customers within those stores.

### Business Example
Selecting several branches for a customer survey.

### AI/ML Use Case
Data may be grouped by hospital, branch, device, or region. Cluster-aware sampling can reduce collection cost when observations are naturally grouped.

This example demonstrates how the concept can be understood and applied.

## Step 3 – Implement the Concept

The following Python code demonstrates the concept.

```python
import pandas as pd

data = pd.DataFrame({
    "customer_id": range(1, 13),
    "store": ["A"]*4 + ["B"]*4 + ["C"]*4
})

selected_clusters = ["A", "C"]

sample = data[data["store"].isin(selected_clusters)]

print(sample)
```

## Interpretation

The sample contains all observations from the selected clusters.

Cluster sampling can be efficient, but selected clusters may not represent the whole population if clusters are very different.

### Key Takeaway
The goal is to understand what the result means and when this concept is useful, not simply to calculate a value.

# 5. Systematic Sampling

## Step 1 – Understand the Concept

### Definition
Systematic sampling selects observations at regular intervals.

If:
N = population size
n = desired sample size

A common sampling interval is:

k = N/n

After choosing a starting point, every kth observation is selected.

## Step 2 – Demonstrate the Concept

### Numerical Example
For N=100 and n=10:

k = 100/10 = 10

If the starting position is 3, select 3, 13, 23, 33, and so on.

### Business Example
Inspecting every 50th product on a production line.

### AI/ML Use Case
Systematic sampling can reduce the amount of data processed when records are ordered appropriately.

This example demonstrates how the concept can be understood and applied.

## Step 3 – Implement the Concept

The following Python code demonstrates the concept.

```python
import numpy as np

population = np.arange(1, 101)

k = 10
start = 3

sample = population[start-1::k]

print("Systematic sample:", sample)
```

## Interpretation

The observations are selected at a fixed interval.

A key risk is periodicity: if the data has a repeating pattern aligned with k, the sample may become biased.

### Key Takeaway
The goal is to understand what the result means and when this concept is useful, not simply to calculate a value.

# 6. Sampling Bias

## Step 1 – Understand the Concept

### Definition
Sampling bias occurs when the sampling process systematically overrepresents or underrepresents certain members or characteristics of the population.

Common causes:
- Convenience sampling
- Undercoverage
- Self-selection
- Non-response
- Incomplete sampling frame

## Step 2 – Demonstrate the Concept

### Numerical Example
If a population contains 80% Group A and 20% Group B, but a sample contains only Group A, the sample does not represent the population.

### Business Example
Surveying only social-media followers may exclude customers who do not use that platform.

### AI/ML Use Case
Biased training data can produce models that perform poorly for underrepresented groups.

This example demonstrates how the concept can be understood and applied.

## Step 3 – Implement the Concept

The following Python code demonstrates the concept.

```python
import pandas as pd

population = pd.DataFrame({
    "group": ["A"] * 80 + ["B"] * 20
})

biased_sample = population[
    population["group"] == "A"
].sample(n=20, random_state=42)

print("Biased sample composition:")
print(biased_sample["group"].value_counts())
```

## Interpretation

The sample contains only Group A, so it cannot represent the full population.

Increasing the sample size does not automatically fix systematic bias.

### Key Takeaway
The goal is to understand what the result means and when this concept is useful, not simply to calculate a value.

# 7. Sampling Error

## Step 1 – Understand the Concept

### Definition
Sampling error is the difference between a population parameter and the corresponding sample statistic caused by observing only part of the population.

Sampling error is expected when sampling is used.

It is different from data-entry errors, measurement errors, or sampling bias.

## Step 2 – Demonstrate the Concept

### Numerical Example
If the population mean is 50 and the sample mean is 48:

Sampling error = 48 − 50 = −2

### Business Example
A survey sample may estimate average customer satisfaction slightly above or below the true population average.

### AI/ML Use Case
Different random samples can produce slightly different estimates of model performance.

This example demonstrates how the concept can be understood and applied.

## Step 3 – Implement the Concept

The following Python code demonstrates the concept.

```python
import numpy as np

rng = np.random.default_rng(42)

population = rng.normal(50, 10, 10000)
sample = rng.choice(population, 100, replace=False)

population_mean = np.mean(population)
sample_mean = np.mean(sample)

sampling_error = sample_mean - population_mean

print("Population mean:", population_mean)
print("Sample mean:", sample_mean)
print("Sampling error:", sampling_error)
```

## Interpretation

The sample mean will generally differ slightly from the population mean.

Increasing the sample size generally reduces sampling variability, assuming the sampling method is appropriate.

### Key Takeaway
The goal is to understand what the result means and when this concept is useful, not simply to calculate a value.

# 8. Sample Size

## Step 1 – Understand the Concept

### Definition
Sample size is the number of observations included in a sample.

A larger sample generally produces more stable estimates, but sample size alone does not guarantee a representative sample.

## Step 2 – Demonstrate the Concept

### Numerical Example
A sample of 20 observations may produce a more variable estimate than a properly selected sample of 2,000 observations.

### Business Example
A company may determine survey sample size based on desired confidence and margin of error.

### AI/ML Use Case
More representative training data can improve generalization, but data quality, coverage, and bias are also important.

This example demonstrates how the concept can be understood and applied.

## Step 3 – Implement the Concept

The following Python code demonstrates the concept.

```python
import numpy as np

rng = np.random.default_rng(42)
population = rng.normal(100, 15, 10000)

for size in [10, 50, 200, 1000]:
    sample = rng.choice(population, size=size, replace=False)
    print(f"Sample size {size}: mean = {np.mean(sample):.2f}")
```

## Interpretation

As sample size increases, the sample estimate generally becomes more stable.

The appropriate sample size depends on variability, confidence level, desired precision, population structure, and cost.

### Key Takeaway
The goal is to understand what the result means and when this concept is useful, not simply to calculate a value.

## 1. Population vs Sample Visualization

A sample should represent the population as closely as possible. The following example shows how a random sample is selected from a larger population.

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)
population = rng.normal(loc=50, scale=10, size=1000)
sample = rng.choice(population, size=50, replace=False)

plt.figure(figsize=(8, 4))
plt.hist(population, bins=30, alpha=0.6, label="Population", edgecolor="black")
plt.hist(sample, bins=15, alpha=0.8, label="Random Sample", edgecolor="black")
plt.title("Population vs Random Sample")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.legend()
plt.show()

print("Population mean:", population.mean())
print("Sample mean:", sample.mean())
```

### Interpretation

The sample contains fewer observations, so its distribution will not exactly match the population. A good random sample should nevertheless provide a reasonable estimate of important population characteristics such as the mean.

## 2. Sampling Error vs Sample Size

Larger samples generally reduce the variability of sample estimates. This experiment repeatedly draws samples and observes how sample means vary.

```python
sample_sizes = [10, 30, 100]
sample_means = {}

for size in sample_sizes:
    means = []
    for _ in range(300):
        sample = rng.choice(population, size=size, replace=False)
        means.append(sample.mean())
    sample_means[size] = means

for size in sample_sizes:
    plt.figure(figsize=(8, 4))
    plt.hist(sample_means[size], bins=20, edgecolor="black")
    plt.title(f"Sampling Distribution of the Mean (n={size})")
    plt.xlabel("Sample mean")
    plt.ylabel("Frequency")
    plt.show()
```

### Interpretation

As sample size increases, the sample means tend to become less variable and concentrate more closely around the population mean. This illustrates why sample size matters when estimating population parameters.

### AI/ML Relevance

Sampling affects training data quality, experiment design, model evaluation, surveys, and statistical inference. Poor sampling can introduce bias and cause a model to perform well on the sample while performing poorly on the broader population.

## 3. Practical Problem-Solving Task

A dataset contains customers from three regions with very different population sizes. A simple random sample may underrepresent a smaller region. A stratified sample can preserve regional representation by sampling from each region separately.

### Decision

Use:
- Random sampling when the population is reasonably homogeneous and every observation should have a similar chance of selection.
- Stratified sampling when important subgroups must be represented.
- Cluster sampling when naturally occurring groups are more practical to sample.
- Systematic sampling when observations have an ordered list and a fixed interval is appropriate.

Always check for sampling bias before using the sample for ML training or statistical inference.
