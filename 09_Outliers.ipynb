## What are Outliers?

### Step 1 – Understand the Concept
An outlier is an observation that is unusually far from the rest of the data. It may represent natural variation, measurement error, data entry error, fraud, or a rare but valid event.

### Step 2 – Demonstrate the Concept
Use the following example to apply the concept to a small dataset.

### Step 3 – Implement the Concept

```python
import numpy as np
data=np.array([10,11,12,12,13,14,15,16,100])
print("Sorted data:", np.sort(data))
```

### Interpretation
The value 100 is much larger than the main cluster and should be investigated rather than automatically deleted.

### Key Takeaway
Outliers require investigation, not automatic removal.

### Step 2 – Demonstrate the Concept
Use the following example to apply the concept to a small dataset.

### Step 3 – Implement the Concept

```python
import pandas as pd
df=pd.DataFrame({"temperature":[20,21,22,23,24,25,45],
                 "hour":[9,10,11,12,13,14,15]})
print(df)
```

## Detecting Outliers

Outliers can be detected using statistical rules, visualizations, domain thresholds, or model-based methods.

### Step 2 – Demonstrate the Concept
Use the following example to apply the concept to a small dataset.

### Step 3 – Implement the Concept

## IQR Method

### Step 1 – Understand the Concept
The IQR is Q3 − Q1. A common rule flags values below Q1−1.5×IQR or above Q3+1.5×IQR.

### Step 2 – Demonstrate the Concept
Use the following example to apply the concept to a small dataset.

### Step 3 – Implement the Concept

```python
import pandas as pd
x=pd.Series([10,11,12,12,13,14,15,16,100])
q1=x.quantile(.25); q3=x.quantile(.75); iqr=q3-q1
lower=q1-1.5*iqr; upper=q3+1.5*iqr
out=x[(x<lower)|(x>upper)]
print(q1,q3,iqr,lower,upper)
print("Outliers:",out.tolist())
```

### Interpretation
The flagged observations lie outside the IQR fences. This method is robust to moderate skew and does not require normality.

### Key Takeaway
IQR is a practical robust method for many skewed datasets.

## Z-Score Method

### Step 1 – Understand the Concept
A Z-score measures how many standard deviations an observation is from the mean. A common rule uses |z| > 3, although the threshold depends on context.

### Step 2 – Demonstrate the Concept
Use the following example to apply the concept to a small dataset.

### Step 3 – Implement the Concept

```python
from scipy.stats import zscore
x=pd.Series([10,11,12,12,13,14,15,16,100])
z=zscore(x)
print(pd.DataFrame({"value":x,"z_score":z}))
print("Potential outliers:", x[abs(z)>3].tolist())
```

### Interpretation
Large absolute Z-scores indicate observations far from the mean. The method is more sensitive to distributional assumptions and extreme values.

### Key Takeaway
Z-score is useful when a mean/standard-deviation based approach is appropriate.

## Box Plot

### Step 1 – Understand the Concept
A box plot summarizes the median, quartiles, spread, and potential outliers using points beyond the whiskers.

### Step 2 – Demonstrate the Concept
Use the following example to apply the concept to a small dataset.

### Step 3 – Implement the Concept

```python
import matplotlib.pyplot as plt
x=[10,11,12,12,13,14,15,16,100]
plt.boxplot(x)
plt.ylabel("Value")
plt.title("Box Plot for Outlier Detection")
plt.show()
```

### Interpretation
The isolated point beyond the upper whisker is a visual indication of a potential outlier.

### Key Takeaway
Box plots provide a compact visual summary of distribution and unusual values.

## Scatter Plot

### Step 1 – Understand the Concept
A scatter plot can reveal unusual observations in the relationship between two variables, including isolated points or unusual patterns.

### Step 2 – Demonstrate the Concept
Use the following example to apply the concept to a small dataset.

### Step 3 – Implement the Concept

```python
x=np.arange(1,10)
y=np.array([2,4,5,7,8,10,11,13,40])
plt.scatter(x,y)
plt.xlabel("X"); plt.ylabel("Y")
plt.title("Scatter Plot with Potential Outlier")
plt.show()
```

### Interpretation
The final observation is far from the main pattern and may influence a fitted model strongly.

### Key Takeaway
Scatter plots are essential when outliers are defined relative to relationships between variables.
