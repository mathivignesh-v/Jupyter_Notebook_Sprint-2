# 1. Introduction to Probability

## Step 1 – Understand the Concept

### Definition
Probability is a mathematical measure of how likely an event is to occur. Its value lies between 0 and 1.

- 0 → impossible
- 1 → certain
- Values between 0 and 1 → different levels of likelihood

For equally likely outcomes:

P(A) = Favorable outcomes / Total possible outcomes

Probability is important in AI/ML because real-world observations contain uncertainty. Models often estimate the probability of a class, event, or future outcome.

## Step 2 – Demonstrate the Concept

### Numerical Example
For a fair die, there are 6 possible outcomes. The probability of rolling a 4 is:

P(4) = 1/6 ≈ 0.1667

### Business Example
If 20 out of 100 leads converted into customers, the observed conversion rate is 20/100 = 0.20.

### AI/ML Use Case
A spam classifier may output P(spam) = 0.92, meaning the model estimates a high probability that the message is spam.

This example demonstrates how the concept can be understood and applied.

## Step 3 – Implement the Concept

The following Python code demonstrates the concept.

```python
import numpy as np

favorable = 1
total = 6

probability = favorable / total

print("Probability of rolling a 4:", probability)
print("Percentage:", probability * 100, "%")
```

## Interpretation

The output is approximately 0.167 or 16.7%.

The code divides the number of favorable outcomes by the total number of equally likely outcomes.

In AI/ML, probability is used for classification confidence, risk scoring, recommendation systems, and uncertainty estimation.

### Key Takeaway
The goal is to understand what the result means and when this concept is useful, not simply to calculate a value.

# 2. Events

## Step 1 – Understand the Concept

### Definition
An event is an outcome or a collection of outcomes that we are interested in.

For a die:

Sample Space = {1, 2, 3, 4, 5, 6}

Event A = getting an even number = {2, 4, 6}

An event can be simple, containing one outcome, or compound, containing multiple outcomes.

## Step 2 – Demonstrate the Concept

### Numerical Example
For a die, the event of getting a number greater than 4 is:

A = {5, 6}

Therefore P(A) = 2/6 = 1/3.

### Business Example
Event A could be "a customer makes a purchase."

### AI/ML Use Case
An event can represent "a transaction is fraudulent" or "a customer churns." 

This example demonstrates how the concept can be understood and applied.

## Step 3 – Implement the Concept

The following Python code demonstrates the concept.

```python
sample_space = {1, 2, 3, 4, 5, 6}
event = {5, 6}

probability = len(event) / len(sample_space)

print("Event:", event)
print("Probability:", probability)
```

## Interpretation

The event contains two favorable outcomes out of six possible outcomes, so the probability is 1/3.

In AI/ML, clearly defining events helps us reason about target outcomes and uncertain model predictions.

### Key Takeaway
The goal is to understand what the result means and when this concept is useful, not simply to calculate a value.

# 3. Sample Space

## Step 1 – Understand the Concept

### Definition
The sample space is the complete set of all possible outcomes of a random experiment.

Examples:

- One coin toss → {H, T}
- One die roll → {1, 2, 3, 4, 5, 6}
- Two coin tosses → {HH, HT, TH, TT}

A complete sample space is important because probability calculations depend on knowing all possible outcomes.

## Step 2 – Demonstrate the Concept

### Numerical Example
For two coin tosses there are four outcomes. Only one outcome is HH.

Therefore:

P(HH) = 1/4 = 0.25

### Business Example
A customer can either renew or not renew a subscription.

### AI/ML Use Case
A binary classification task has two possible classes, while multiclass classification has multiple possible class outcomes.

This example demonstrates how the concept can be understood and applied.

## Step 3 – Implement the Concept

The following Python code demonstrates the concept.

```python
import itertools

sample_space = list(itertools.product(["H", "T"], repeat=2))

print("Sample Space:", sample_space)
print("Number of outcomes:", len(sample_space))

favorable = [x for x in sample_space if x == ("H", "H")]
print("P(HH):", len(favorable) / len(sample_space))
```

## Interpretation

The sample space contains HH, HT, TH, and TT. HH is one of four equally likely outcomes, so its probability is 0.25.

In AI/ML, defining the possible target classes correctly is an important first step in classification problems.

### Key Takeaway
The goal is to understand what the result means and when this concept is useful, not simply to calculate a value.

# 4. Independent Events

## Step 1 – Understand the Concept

### Definition
Two events are independent when the occurrence of one event does not change the probability of the other.

For independent events:

P(A ∩ B) = P(A) × P(B)

Independence is an assumption about the relationship between events; it should not be assumed without justification.

## Step 2 – Demonstrate the Concept

### Numerical Example
If P(A) = 0.6 and P(B) = 0.4 and they are independent:

P(A ∩ B) = 0.6 × 0.4 = 0.24

### Business Example
Two unrelated random selections may be treated as independent under suitable conditions.

### AI/ML Use Case
Naive Bayes assumes features are conditionally independent given the class.

This example demonstrates how the concept can be understood and applied.

## Step 3 – Implement the Concept

The following Python code demonstrates the concept.

```python
p_a = 0.60
p_b = 0.40

p_intersection = p_a * p_b

print("P(A and B):", p_intersection)
```

## Interpretation

The result is 0.24.

The multiplication rule works directly because the events are assumed to be independent.

In AI/ML, independence assumptions can make probabilistic models simpler and computationally efficient.

### Key Takeaway
The goal is to understand what the result means and when this concept is useful, not simply to calculate a value.

# 5. Dependent Events

## Step 1 – Understand the Concept

### Definition
Two events are dependent when the occurrence of one event changes the probability of another.

General multiplication rule:

P(A ∩ B) = P(A) × P(B|A)

Sampling without replacement is a common example because removing one item changes the remaining population.

## Step 2 – Demonstrate the Concept

### Numerical Example
A bag contains 3 red and 2 blue balls.

P(first red) = 3/5

After one red is removed:

P(second red | first red) = 2/4

Therefore:

P(two red) = 3/5 × 2/4 = 0.30

### Business Example
A customer's previous purchase can affect the probability of a future purchase.

### AI/ML Use Case
Relationships between features can create dependence, which matters when selecting statistical assumptions and models.

This example demonstrates how the concept can be understood and applied.

## Step 3 – Implement the Concept

The following Python code demonstrates the concept.

```python
red = 3
total = 5

p_first_red = red / total
p_second_red_given_first = (red - 1) / (total - 1)

p_two_red = p_first_red * p_second_red_given_first

print("P(first red):", p_first_red)
print("P(second red | first red):", p_second_red_given_first)
print("P(two red):", p_two_red)
```

## Interpretation

The second probability changes because the first draw changes the composition of the bag.

In AI/ML, understanding dependence is important when interpreting correlations, conditional relationships, and probabilistic models.

### Key Takeaway
The goal is to understand what the result means and when this concept is useful, not simply to calculate a value.

# 6. Conditional Probability

## Step 1 – Understand the Concept

### Definition
Conditional probability is the probability of one event occurring given that another event has already occurred.

Formula:

P(A|B) = P(A ∩ B) / P(B)

The condition changes the reference group from the complete population to the group represented by B.

## Step 2 – Demonstrate the Concept

### Numerical Example
Suppose 30% of customers use a mobile application and 12% both use the application and purchase.

P(purchase | app) = 0.12 / 0.30 = 0.40

### Business Example
A company may calculate the probability of purchase given that a customer received a promotion.

### AI/ML Use Case
Classification can be viewed as estimating P(class | features).

This example demonstrates how the concept can be understood and applied.

## Step 3 – Implement the Concept

The following Python code demonstrates the concept.

```python
p_app = 0.30
p_app_and_purchase = 0.12

p_purchase_given_app = p_app_and_purchase / p_app

print("P(purchase | app):", p_purchase_given_app)
print("Percentage:", p_purchase_given_app * 100, "%")
```

## Interpretation

The output is 0.40, meaning the estimated probability of purchase among app users is 40%.

Conditional probability is a foundation of probabilistic classification and Bayesian reasoning.

### Key Takeaway
The goal is to understand what the result means and when this concept is useful, not simply to calculate a value.

# 7. Bayes' Theorem

## Step 1 – Understand the Concept

### Definition
Bayes' theorem updates the probability of a hypothesis after observing evidence.

Formula:

P(A|B) = [P(B|A) × P(A)] / P(B)

Where:
- P(A) = prior probability
- P(B|A) = likelihood
- P(B) = evidence probability
- P(A|B) = posterior probability

It combines prior knowledge with new evidence.

## Step 2 – Demonstrate the Concept

### Numerical Example
Suppose:
- Disease prevalence = 1%
- Positive test given disease = 90%
- Positive test given no disease = 5%

A positive result does not automatically mean a 90% chance of disease. The prior probability must also be considered.

### Business Example
A fraud detection system can update the probability of fraud after observing transaction characteristics.

### AI/ML Use Case
Bayesian methods are used in spam filtering, diagnosis, fraud detection, Bayesian classifiers, and probabilistic inference.

This example demonstrates how the concept can be understood and applied.

## Step 3 – Implement the Concept

The following Python code demonstrates the concept.

```python
p_disease = 0.01
p_positive_given_disease = 0.90
p_positive_given_no_disease = 0.05

p_positive = (
    p_positive_given_disease * p_disease
    + p_positive_given_no_disease * (1 - p_disease)
)

posterior = (
    p_positive_given_disease * p_disease
    / p_positive
)

print("P(disease | positive):", posterior)
print("Percentage:", posterior * 100, "%")
```

## Interpretation

The posterior probability is about 15.38%.

The important point is that evidence must be interpreted together with the prior probability.

Bayesian reasoning is widely used when an AI/ML system must update beliefs as new information becomes available.

### Key Takeaway
The goal is to understand what the result means and when this concept is useful, not simply to calculate a value.

# 8. Probability Rules

## Step 1 – Understand the Concept

### Definition
Important probability rules include:

Complement:
P(Aᶜ) = 1 − P(A)

Addition:
P(A ∪ B) = P(A) + P(B) − P(A ∩ B)

Multiplication:
P(A ∩ B) = P(A)P(B|A)

For independent events:
P(A ∩ B) = P(A)P(B)

These rules help combine individual probability statements.

## Step 2 – Demonstrate the Concept

### Numerical Example
Suppose P(A)=0.6, P(B)=0.5 and P(A∩B)=0.3.

Then:

P(A∪B)=0.6+0.5−0.3=0.8

And:

P(Aᶜ)=1−0.6=0.4

### Business Example
A company can calculate the probability that a customer either clicks an advertisement or makes a purchase.

### AI/ML Use Case
These rules support probabilistic reasoning and risk calculations.

This example demonstrates how the concept can be understood and applied.

## Step 3 – Implement the Concept

The following Python code demonstrates the concept.

```python
p_a = 0.60
p_b = 0.50
p_intersection = 0.30

p_union = p_a + p_b - p_intersection
p_not_a = 1 - p_a

print("P(A or B):", p_union)
print("P(not A):", p_not_a)
```

## Interpretation

The union is 0.8 and the complement of A is 0.4.

The subtraction of the intersection prevents overlapping outcomes from being counted twice.

### Key Takeaway
The goal is to understand what the result means and when this concept is useful, not simply to calculate a value.

# 9. Random Variables

### Definition
A random variable assigns a numerical value to the outcome of a random experiment.

Discrete random variables have countable values, such as number of purchases.

Continuous random variables can take values across an interval, such as waiting time.

Random variables provide a mathematical way to represent uncertain quantities.

# 10. Expected Value

## Step 1 – Understand the Concept

### Definition
Expected value is the probability-weighted average of all possible outcomes.

Formula:

E[X] = Σ xP(X=x)

Expected value represents the long-run average outcome. It does not have to be one of the outcomes that can actually occur.

## Step 2 – Demonstrate the Concept

### Numerical Example
Suppose:
- ₹100 profit with probability 0.6
- ₹0 profit with probability 0.4

E[X] = 100(0.6) + 0(0.4) = ₹60

### Business Example
Expected revenue can be calculated by weighting possible revenue outcomes by their probabilities.

### AI/ML Use Case
Expected loss is a core idea in statistical learning and model optimization.

This example demonstrates how the concept can be understood and applied.

## Step 3 – Implement the Concept

The following Python code demonstrates the concept.

```python
outcomes = [100, 0]
probabilities = [0.60, 0.40]

expected_value = sum(
    x * p for x, p in zip(outcomes, probabilities)
)

print("Expected value: ₹", expected_value)
```

## Interpretation

The expected value is ₹60.

The result represents the average outcome expected over many repetitions under the given probabilities.

In AI/ML, expected loss helps define what a model is optimized to minimize.

### Key Takeaway
The goal is to understand what the result means and when this concept is useful, not simply to calculate a value.

## 1. Probability Visualization

## 2. Practical AI/ML Problem – Conditional Probability

A fraud detection model flags 8% of transactions. Among genuinely fraudulent transactions, the model flags 90%. Among legitimate transactions, it incorrectly flags 5%.

Use Bayes' theorem to estimate the probability that a transaction is actually fraudulent when the model flags it.

Let:
- `F` = transaction is fraudulent
- `A` = model flags the transaction

`P(F) = 0.08`, `P(A|F) = 0.90`, and `P(A|not F) = 0.05`.

```python
p_fraud = 0.08
p_flag_given_fraud = 0.90
p_flag_given_legitimate = 0.05

p_legitimate = 1 - p_fraud
p_flag = (p_flag_given_fraud * p_fraud) + (p_flag_given_legitimate * p_legitimate)

p_fraud_given_flag = (p_flag_given_fraud * p_fraud) / p_flag

print("P(flag):", p_flag)
print("P(fraud | flag):", p_fraud_given_flag)
```

### Interpretation

The probability that a flagged transaction is genuinely fraudulent is different from the model's sensitivity. Bayes' theorem combines the model's conditional probabilities with the base rate of fraud. This is an important concept when interpreting classification systems.

### AI/ML Relevance

Conditional probability and Bayes' theorem are fundamental to classification, medical diagnosis, fraud detection, recommendation systems, and probabilistic machine learning.
