# 1. Bernoulli Distribution

## Step 1 – Understand the Concept

### Definition
The Bernoulli distribution represents a single trial with exactly two possible outcomes.

Usually:
- X = 1 → success
- X = 0 → failure

It is the simplest discrete probability distribution and forms the foundation of the Binomial distribution.

## Step 2 – Demonstrate the Concept

### Formula / Key Idea
P(X=x) = p^x(1-p)^(1-x), x ∈ {0,1}

Where:
- p = probability of success
- 1-p = probability of failure

Mean = p
Variance = p(1-p)

### Where It Is Used
Use it for one binary trial.

### Real-World Example
A customer either purchases or does not purchase.

### AI/ML Example
A binary target such as spam/not spam can be represented using a Bernoulli random variable.

This example demonstrates how the concept can be understood and applied.

## Step 3 – Implement the Concept

The following Python code demonstrates the concept.

```python
import numpy as np
import matplotlib.pyplot as plt

p = 0.70
x = np.array([0, 1])
probabilities = np.array([1-p, p])

plt.bar(x, probabilities)
plt.xticks(x, ["Failure (0)", "Success (1)"])
plt.xlabel("Outcome")
plt.ylabel("Probability")
plt.title("Bernoulli Distribution")
plt.show()
```

## Interpretation

The success bar is 0.70 and the failure bar is 0.30.

The visualization shows that there are only two possible outcomes.

AI/ML connection: Binary classification targets naturally have two possible states.

### Key Takeaway
The goal is to understand what the result means and when this concept is useful, not simply to calculate a value.

# 2. Binomial Distribution

## Step 1 – Understand the Concept

### Definition
The Binomial distribution represents the number of successes in a fixed number n of independent Bernoulli trials.

Conditions:
- Fixed number of trials
- Two outcomes per trial
- Independent trials
- Constant probability of success

## Step 2 – Demonstrate the Concept

### Formula / Key Idea
P(X=k) = C(n,k)p^k(1-p)^(n-k)

Where:
- n = number of trials
- k = number of successes
- p = success probability

Mean = np
Variance = np(1-p)

### Where It Is Used
Counting successes across repeated binary trials.

### Real-World Example
Number of customers who purchase among 20 contacted customers.

### AI/ML Example
Modeling the number of positive outcomes among repeated binary experiments.

This example demonstrates how the concept can be understood and applied.

## Step 3 – Implement the Concept

The following Python code demonstrates the concept.

```python
import numpy as np
import matplotlib.pyplot as plt
from math import comb

n = 20
p = 0.30

k = np.arange(n + 1)
probabilities = np.array([
    comb(n, i) * p**i * (1-p)**(n-i)
    for i in k
])

plt.bar(k, probabilities)
plt.xlabel("Number of successes")
plt.ylabel("Probability")
plt.title("Binomial Distribution")
plt.show()
```

## Interpretation

Each bar represents the probability of getting a particular number of successes.

Increasing p shifts probability toward larger success counts.

AI/ML connection: It is useful when a binary event is repeated a fixed number of times.

### Key Takeaway
The goal is to understand what the result means and when this concept is useful, not simply to calculate a value.

# 3. Poisson Distribution

## Step 1 – Understand the Concept

### Definition
The Poisson distribution models the number of times an event occurs during a fixed interval when events occur at an average rate λ.

It is commonly used for count data.

## Step 2 – Demonstrate the Concept

### Formula / Key Idea
P(X=k) = (e^-λ λ^k) / k!

Where:
- λ = average event rate
- k = number of observed events

Mean = λ
Variance = λ

### Where It Is Used
Counts of events per time, distance, area, or volume.

### Real-World Example
Number of customer support calls received per hour.

### AI/ML Example
Modeling incidents, requests, arrivals, failures, or other event counts.

This example demonstrates how the concept can be understood and applied.

## Step 3 – Implement the Concept

The following Python code demonstrates the concept.

```python
import numpy as np
import matplotlib.pyplot as plt
from math import exp, factorial

lam = 4
k = np.arange(0, 15)

probabilities = np.array([
    (exp(-lam) * lam**i) / factorial(i)
    for i in k
])

plt.bar(k, probabilities)
plt.xlabel("Number of events")
plt.ylabel("Probability")
plt.title("Poisson Distribution")
plt.show()
```

## Interpretation

The distribution is centered around the average rate λ=4.

An important property is that the mean and variance are both λ.

AI/ML connection: Count-event problems can often be explored using Poisson models.

### Key Takeaway
The goal is to understand what the result means and when this concept is useful, not simply to calculate a value.

# 4. Uniform Distribution

## Step 1 – Understand the Concept

### Definition
A continuous Uniform distribution gives equal probability density to every value within a specified interval [a,b].

Unlike the Normal distribution, it does not favor values near the center.

## Step 2 – Demonstrate the Concept

### Formula / Key Idea
f(x) = 1/(b-a), for a ≤ x ≤ b

Mean = (a+b)/2
Variance = (b-a)^2/12

### Where It Is Used
Simulation, random sampling, and situations where values in a range are equally plausible.

### Real-World Example
Selecting a random time uniformly from a defined interval.

### AI/ML Example
Simulation and sampling for selected parameter ranges.

This example demonstrates how the concept can be understood and applied.

## Step 3 – Implement the Concept

The following Python code demonstrates the concept.

```python
import numpy as np
import matplotlib.pyplot as plt

a, b = 0, 10
rng = np.random.default_rng(42)
data = rng.uniform(a, b, 2000)

plt.hist(data, bins=25)
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Uniform Distribution")
plt.show()
```

## Interpretation

The histogram should be relatively flat from 0 to 10.

AI/ML connection: Uniform random sampling is useful when all values within a defined range are treated similarly.

### Key Takeaway
The goal is to understand what the result means and when this concept is useful, not simply to calculate a value.

# 5. Normal Distribution

## Step 1 – Understand the Concept

### Definition
The Normal distribution is a continuous, symmetric, bell-shaped distribution described by its mean μ and standard deviation σ.

For a perfect Normal distribution:

Mean = Median = Mode

## Step 2 – Demonstrate the Concept

### Formula / Key Idea
f(x) = [1/(σ√(2π))] exp[-(x-μ)^2/(2σ²)]

Where:
- μ = mean
- σ = standard deviation

Empirical rule:
- About 68% within ±1σ
- About 95% within ±2σ
- About 99.7% within ±3σ

### Where It Is Used
Continuous measurements that are approximately symmetric and bell-shaped.

### Real-World Example
Some measurement errors and physical measurements.

### AI/ML Example
Gaussian Naive Bayes, noise models, residual analysis, statistical inference, and probabilistic modeling.

This example demonstrates how the concept can be understood and applied.

## Step 3 – Implement the Concept

The following Python code demonstrates the concept.

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)

mu = 50
sigma = 10

data = rng.normal(mu, sigma, 3000)

plt.hist(data, bins=35)
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Normal Distribution")
plt.show()
```

## Interpretation

The histogram should show a bell-shaped pattern centered near 50.

Changing sigma changes the spread, while changing mu shifts the center.

AI/ML connection: Understanding normality helps interpret standardized values and statistical assumptions, although ML algorithms do not universally require normal features.

### Key Takeaway
The goal is to understand what the result means and when this concept is useful, not simply to calculate a value.

# 6. Exponential Distribution

## Step 1 – Understand the Concept

### Definition
The Exponential distribution models the waiting time until the next event in a Poisson process.

It is continuous and takes only non-negative values.

## Step 2 – Demonstrate the Concept

### Formula / Key Idea
f(x) = λe^(-λx), x ≥ 0

Where:
- λ = event rate

Mean = 1/λ
Variance = 1/λ²

### Where It Is Used
Waiting-time and time-between-event problems.

### Real-World Example
Time until the next customer arrives at a service counter.

### AI/ML Example
Reliability analysis, service waiting times, event-arrival modeling, and survival-related applications.

This example demonstrates how the concept can be understood and applied.

## Step 3 – Implement the Concept

The following Python code demonstrates the concept.

```python
import numpy as np
import matplotlib.pyplot as plt

rate = 0.5

rng = np.random.default_rng(42)
data = rng.exponential(scale=1/rate, size=3000)

plt.hist(data, bins=35)
plt.xlabel("Waiting Time")
plt.ylabel("Frequency")
plt.title("Exponential Distribution")
plt.show()
```

## Interpretation

The distribution has a strong right tail. Short waiting times are more common than very long waiting times.

AI/ML connection: The distribution is useful when modeling time between independent events occurring at a constant average rate.

### Key Takeaway
The goal is to understand what the result means and when this concept is useful, not simply to calculate a value.

## 1. Comparing Discrete Probability Distributions

Bernoulli and Binomial distributions are related but represent different situations. Bernoulli models one binary trial, while Binomial models the number of successes across a fixed number of independent Bernoulli trials.

```python
import numpy as np
import matplotlib.pyplot as plt
from math import comb

p = 0.6
n = 10
x = np.arange(0, n + 1)
binomial_pmf = np.array([comb(n, k) * p**k * (1-p)**(n-k) for k in x])

plt.figure(figsize=(8, 4))
plt.bar(x, binomial_pmf, edgecolor="black")
plt.title("Binomial Distribution: n=10, p=0.6")
plt.xlabel("Number of successes")
plt.ylabel("Probability")
plt.xticks(x)
plt.show()
```

### Interpretation

The Binomial distribution gives the probability of obtaining each possible number of successes in 10 independent trials. The highest probabilities occur near the expected number of successes, `n × p = 6`.

## 2. Practical AI/ML Problem – Event Counts

Suppose an application receives an average of 4 error events per minute. A Poisson distribution can model the probability of observing exactly 6 errors in one minute.

`P(X=k) = (e^-λ λ^k) / k!`

```python
import math

lam = 4
k = 6

poisson_probability = (math.exp(-lam) * lam**k) / math.factorial(k)

print("P(X = 6):", poisson_probability)
```

### Interpretation

The Poisson model is appropriate when the task concerns the count of events occurring in a fixed interval and the average event rate is known. In practice, the assumptions should be checked before using the model.

### AI/ML Relevance

Probability distributions are used to model binary outcomes, counts, continuous measurements, event arrivals, uncertainty, and likelihoods. Selecting a suitable distribution helps engineers build realistic statistical models and understand data-generating processes.
