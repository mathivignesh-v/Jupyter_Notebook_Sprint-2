## Correlation Matrix

### Step 1 – Understand the Concept
Summarizes pairwise numerical relationships and is an initial redundancy-screening tool.

### Step 2 – Demonstrate the Concept
Inspect relationships among age, income, spending, and visits.

```python
import pandas as pd, seaborn as sns, matplotlib.pyplot as plt
df=pd.DataFrame({"age":[22,25,30,35,40,45,50,55],"income":[30,35,42,48,55,62,70,78],"spend":[20,24,30,35,42,47,53,60],"visits":[3,4,5,6,7,8,9,10]})
corr=df.corr()
sns.heatmap(corr,annot=True,cmap="coolwarm",center=0); plt.show()
```

### Interpretation
Strong feature-feature correlations warrant further investigation.

### Key Takeaway
Correlation is an EDA signal, not proof of redundancy.

## Multicollinearity

### Step 1 – Understand the Concept
Occurs when predictors contain strong linear dependencies, making regression coefficients unstable or hard to interpret.

### Step 2 – Demonstrate the Concept
Income and annual salary can encode overlapping information.

```python
print(corr)
print("Inspect highly correlated predictor pairs before regression.")
```

### Interpretation
High predictor correlation can indicate multicollinearity.

### Key Takeaway
Investigate multicollinearity before interpreting individual regression coefficients.

## Variance Inflation Factor (VIF)

### Step 1 – Understand the Concept
VIF quantifies how much coefficient variance is inflated by linear relationships among predictors.

### Step 2 – Demonstrate the Concept
Calculate VIF for a regression feature set.

```python
from statsmodels.stats.outliers_influence import variance_inflation_factor
from statsmodels.tools.tools import add_constant
X=add_constant(df[["age","income","spend","visits"]])
print(pd.DataFrame({"feature":X.columns,"VIF":[variance_inflation_factor(X.values,i) for i in range(X.shape[1])]}))
```

### Interpretation
Large VIF values are warning signals; there is no universally correct cutoff.

### Key Takeaway
Use VIF with correlation, domain knowledge, and model purpose.
