# 1. Range

## Step 1 – Understand the Concept

### Definition
Range measures the total spread from the smallest value to the largest value in a dataset.

### Formula / Key Idea
Range = Maximum − Minimum


## Step 2 – Demonstrate the Concept

### Numerical Example
For 10, 15, 20, 25, 30: Range = 30 − 10 = 20.

### Business Example
A company can compare the range of daily sales to understand how much sales vary between the lowest and highest days.

### AI/ML Use Case
Range gives a quick view of the spread of a numerical feature during exploratory data analysis.

This example demonstrates how the concept helps us understand the spread or distribution of data.


## Step 3 – Implement the Concept

The following Python code demonstrates the concept.

```python
import numpy as np

data = np.array([10, 15, 20, 25, 30])
range_value = np.max(data) - np.min(data)
print('Range:', range_value)
```

## Interpretation

A range of 20 means the largest value is 20 units above the smallest value. It is simple but can be strongly affected by extreme values.

### Key Takeaway
The goal is to understand what the result means and when this concept is useful, not simply to calculate a value.


# 2. Variance

## Step 1 – Understand the Concept

### Definition
Variance measures how far observations are spread from their mean by calculating the average squared deviation from the mean.

### Formula / Key Idea
Population variance = Σ(x − μ)² / N


## Step 2 – Demonstrate the Concept

### Numerical Example
For 2, 4, 6: mean = 4; squared deviations are 4, 0, 4; population variance = 8/3 ≈ 2.67.

### Business Example
A business can compare variance in delivery times to identify processes with inconsistent performance.

### AI/ML Use Case
Variance helps quantify feature spread and is fundamental to many statistical and ML methods.

This example demonstrates how the concept helps us understand the spread or distribution of data.


## Step 3 – Implement the Concept

The following Python code demonstrates the concept.

```python
import numpy as np

data = np.array([2, 4, 6], dtype=float)
variance = np.var(data)
print('Population variance:', variance)
```

## Interpretation

A larger variance means observations are more spread out around the mean. Squaring deviations makes the measure sensitive to large deviations.

### Key Takeaway
The goal is to understand what the result means and when this concept is useful, not simply to calculate a value.


# 3. Standard Deviation

## Step 1 – Understand the Concept

### Definition
Standard deviation measures the typical spread of observations around the mean and is the square root of variance.

### Formula / Key Idea
Population SD = √[Σ(x − μ)² / N]


## Step 2 – Demonstrate the Concept

### Numerical Example
For 2, 4, 6, population variance is about 2.67, so standard deviation is √2.67 ≈ 1.63.

### Business Example
A company can use standard deviation to understand how consistently employees meet daily targets.

### AI/ML Use Case
Standard deviation is widely used in feature scaling, data analysis, anomaly detection, and understanding model inputs.

This example demonstrates how the concept helps us understand the spread or distribution of data.


## Step 3 – Implement the Concept

The following Python code demonstrates the concept.

```python
import numpy as np

data = np.array([2, 4, 6], dtype=float)
std = np.std(data)
print('Population standard deviation:', std)
```

## Interpretation

The standard deviation is about 1.63, indicating the typical distance of observations from the mean is around that size. It is in the same unit as the original data.

### Key Takeaway
The goal is to understand what the result means and when this concept is useful, not simply to calculate a value.


# 4. Quartiles

## Step 1 – Understand the Concept

### Definition
Quartiles divide ordered data into four parts. Q1 is the 25th percentile, Q2 is the median, and Q3 is the 75th percentile.

### Formula / Key Idea
Q1 = 25th percentile; Q2 = 50th percentile; Q3 = 75th percentile


## Step 2 – Demonstrate the Concept

### Numerical Example
For 10, 20, 30, 40, 50, 60, 70: Q2 is 40. Using NumPy's percentile method, Q1 = 25 and Q3 = 55.

### Business Example
A company can divide customer spending into quartiles to identify low, middle, and high-spending groups.

### AI/ML Use Case
Quartiles help understand feature distributions and are useful for detecting outliers with IQR.

This example demonstrates how the concept helps us understand the spread or distribution of data.


## Step 3 – Implement the Concept

The following Python code demonstrates the concept.

```python
import numpy as np

data = np.array([10, 20, 30, 40, 50, 60, 70])
q1, q2, q3 = np.percentile(data, [25, 50, 75])
print('Q1:', q1)
print('Q2:', q2)
print('Q3:', q3)
```

## Interpretation

The quartiles show where the lower 25%, middle 50%, and upper 25% of observations lie.

### Key Takeaway
The goal is to understand what the result means and when this concept is useful, not simply to calculate a value.


# 5. Percentiles

## Step 1 – Understand the Concept

### Definition
A percentile indicates the value below which a given percentage of observations falls.

### Formula / Key Idea
The pth percentile is the value below which approximately p% of observations lie.


## Step 2 – Demonstrate the Concept

### Numerical Example
If a student's score is at the 90th percentile, the student scored as high as or higher than approximately 90% of the reference group.

### Business Example
An e-commerce company can use the 95th percentile of delivery time to understand a high-end customer waiting time.

### AI/ML Use Case
Percentiles are useful for threshold selection, anomaly detection, feature analysis, and understanding skewed data.

This example demonstrates how the concept helps us understand the spread or distribution of data.


## Step 3 – Implement the Concept

The following Python code demonstrates the concept.

```python
import numpy as np

data = np.array([50, 60, 65, 70, 75, 80, 85, 90, 95])
p90 = np.percentile(data, 90)
print('90th percentile:', p90)
```

## Interpretation

Percentiles provide a position-based view of data and are useful when a fixed percentage threshold is more meaningful than an average.

### Key Takeaway
The goal is to understand what the result means and when this concept is useful, not simply to calculate a value.


# 6. Interquartile Range (IQR)

## Step 1 – Understand the Concept

### Definition
The interquartile range measures the spread of the middle 50% of a dataset and is the difference between Q3 and Q1.

### Formula / Key Idea
IQR = Q3 − Q1


## Step 2 – Demonstrate the Concept

### Numerical Example
If Q1 = 25 and Q3 = 55, then IQR = 55 − 25 = 30.

### Business Example
A business can use IQR to understand the typical spread of customer spending while reducing the influence of unusually large purchases.

### AI/ML Use Case
IQR is commonly used for outlier detection, often using lower and upper fences based on 1.5 × IQR.

This example demonstrates how the concept helps us understand the spread or distribution of data.


## Step 3 – Implement the Concept

The following Python code demonstrates the concept.

```python
import numpy as np

data = np.array([10, 20, 30, 40, 50, 60, 70])
q1, q3 = np.percentile(data, [25, 75])
iqr = q3 - q1
print('Q1:', q1)
print('Q3:', q3)
print('IQR:', iqr)
```

## Interpretation

An IQR of 30 means the middle 50% of observations span 30 units. It is more resistant to extreme values than the range.

### Key Takeaway
The goal is to understand what the result means and when this concept is useful, not simply to calculate a value.


# 7. Coefficient of Variation

## Step 1 – Understand the Concept

### Definition
The coefficient of variation compares standard deviation with the mean. It expresses relative variability, usually as a percentage.

### Formula / Key Idea
CV = (Standard Deviation / Mean) × 100


## Step 2 – Demonstrate the Concept

### Numerical Example
If mean = 50 and standard deviation = 5, CV = (5/50) × 100 = 10%.

### Business Example
A company can compare variability of two products even when their average sales levels are different.

### AI/ML Use Case
CV can help compare relative variation between numerical features measured on different scales, when the mean is meaningful and non-zero.

This example demonstrates how the concept helps us understand the spread or distribution of data.


## Step 3 – Implement the Concept

The following Python code demonstrates the concept.

```python
import numpy as np

data = np.array([45, 48, 50, 52, 55], dtype=float)
mean = np.mean(data)
std = np.std(data)
cv = (std / mean) * 100
print('Mean:', mean)
print('Standard deviation:', std)
print('Coefficient of variation (%):', cv)
```

## Interpretation

A lower CV means less relative variability, while a higher CV means more variability relative to the mean.

### Key Takeaway
The goal is to understand what the result means and when this concept is useful, not simply to calculate a value.
