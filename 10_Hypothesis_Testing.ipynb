## Null Hypothesis

### Step 1 – Understand the Concept
H0 is the default claim, commonly no difference/effect/association.

### Step 2 – Demonstrate the Concept
An A/B test may define H0 as equal conversion rates.

```python
import numpy as np
from scipy import stats
control=[10,11,12,10,13,12]
test=[11,13,14,12,15,13]
print(stats.ttest_ind(control,test,equal_var=False))
```

### Interpretation
The test result is evaluated against a preselected decision rule.

### Key Takeaway
H0 is the baseline claim tested.

## Alternative Hypothesis

### Step 1 – Understand the Concept
H1 represents the difference, effect, or association being investigated.

### Step 2 – Demonstrate the Concept
A new webpage may be hypothesized to have a different conversion rate.

```python
alpha=0.05
print("Reject H0 when p-value <",alpha)
```

### Interpretation
The alternative is supported only when evidence is sufficiently inconsistent with H0.

### Key Takeaway
H1 is the competing claim.

## p-value

### Step 1 – Understand the Concept
The p-value measures how unusual the observed test statistic, or something more extreme, would be if H0 were true.

### Step 2 – Demonstrate the Concept
A small p-value indicates stronger evidence against H0 under the test assumptions.

```python
p=0.02
alpha=0.05
print("Reject H0" if p<alpha else "Fail to reject H0")
```

### Interpretation
A p-value below alpha leads to rejection of H0 under the chosen procedure.

### Key Takeaway
A p-value is not the probability that H0 is true.

## Confidence Interval

### Step 1 – Understand the Concept
A confidence interval gives a range produced by a method designed to capture the population parameter at a stated long-run rate.

### Step 2 – Demonstrate the Concept
Estimate a mean wait time with a 95% interval.

```python
x=np.array([8,10,9,11,12,10,9,13,11,10])
mean=x.mean(); se=stats.sem(x)
print(stats.t.interval(.95,len(x)-1,loc=mean,scale=se))
```

### Interpretation
The interval communicates estimate uncertainty.

### Key Takeaway
Confidence intervals add uncertainty information to point estimates.

## Significance Level

### Step 1 – Understand the Concept
Alpha is the preselected threshold for statistical significance, often 0.05.

### Step 2 – Demonstrate the Concept
An experiment can set alpha before data analysis.

```python
alpha=0.05
print("Alpha:",alpha)
```

### Interpretation
Alpha defines the decision threshold for rejecting H0.

### Key Takeaway
Choose alpha before inspecting the result.

## Type I Error

### Step 1 – Understand the Concept
Rejecting a true H0 is a Type I error (false positive).

### Step 2 – Demonstrate the Concept
A fraud system flags a legitimate transaction.

```python
alpha=0.05
print("Illustrative Type I error target:",alpha)
```

### Interpretation
Lower alpha generally reduces the allowed false-positive rate under repeated use of the procedure.

### Key Takeaway
Type I = false positive.

## Type II Error

### Step 1 – Understand the Concept
Failing to reject a false H0 is a Type II error (false negative).

### Step 2 – Demonstrate the Concept
A quality test misses a real process improvement.

```python
beta=.20
print("Illustrative beta:",beta,"Power:",1-beta)
```

### Interpretation
Power is 1−β for a specified alternative and study design.

### Key Takeaway
Type II = false negative.

## One-tailed Test

### Step 1 – Understand the Concept
A one-tailed test uses a directional alternative such as greater than or less than.

### Step 2 – Demonstrate the Concept
Test whether a new process reduces average defects.

```python
x=np.array([8,9,7,8,6,7,8])
t,p2=stats.ttest_1samp(x,9)
p1=p2/2 if t<0 else 1-p2/2
print(t,p1)
```

### Interpretation
The direction must be specified before looking at the result.

### Key Takeaway
Use only when the research question is genuinely directional.

## Two-tailed Test

### Step 1 – Understand the Concept
A two-tailed test detects differences in either direction.

### Step 2 – Demonstrate the Concept
Test whether average delivery time differs from a target.

```python
x=np.array([8,9,7,8,6,7,8])
print(stats.ttest_1samp(x,9))
```

### Interpretation
A small two-tailed p-value indicates evidence of a difference from the hypothesized value.

### Key Takeaway
Use when both directions matter.
